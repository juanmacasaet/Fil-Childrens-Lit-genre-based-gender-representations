## Replace accented characters with ascii

```python
import unicodedata

def remove_accents(text):
    if not isinstance(text, str):
        return text
    nfkd_form = unicodedata.normalize('NFKD', text)
    return nfkd_form.encode('ascii', 'ignore').decode('utf-8')
```

```python
all_dfs['sentence_clean'] = all_dfs['sentence'].apply(remove_accents)
```

```python
all_dfs['genre1'].value_counts()
```

```python
all_annotations = []

paragraph_id = all_dfs['sentence_id'].tolist()
texts = all_dfs["sentence_clean"].tolist()
filename = all_dfs['filename'].tolist()
genres1 = all_dfs["genre1"].tolist()
genres2 = all_dfs["genre2"].tolist()
genre_group = all_dfs['genre_grp1'].tolist()

for _, (doc, paragraph_id, filename, genre1, genre2, genre_group) in enumerate(
    zip(nlp.pipe(texts, batch_size=64), paragraph_id, filename, genres1, genres2, genre_group
       )):
    for token in doc:
        all_annotations.append({
            "filename": filename,
            "sentence_id": paragraph_id,
            "token": token.text,
            "pos": token.pos_,
            "entity": token.ent_type_ or "0",
            "genre1": genre1,
            'genre2': genre2,
            "genre_group": genre_group
        })

annotations_df = pd.DataFrame(all_annotations)
```

## Insert corrected POS

```python
POS_ref = pd.read_csv('genrePOSv2-updated.csv')
POS_ref = POS_ref[['token_low', 'pos']]
POS_ref
```

```python
token_pos = token_pos[['filename', 'sentence_id', 'token', 'token_low', 'pos', 'entity', 'genre1', 'genre2', 'genre_group']]
token_pos
```

```python
token_pos_copy = token_pos.copy()

merged = token_pos_copy.merge(
    POS_ref,
    on=['token_low', 'pos'],
    how='left',
    indicator=True
)
pos_lookup = POS_ref.groupby("token_low")["pos"].apply(list)

def fix_pos(row):
    word = row["token_low"]
    current_pos = row["pos"]
    
    if word not in pos_lookup:
        return current_pos  # no correction available
    
    valid_pos = pos_lookup[word]
    
    # Case 2: incorrect POS → replace
    if current_pos not in valid_pos:
        return valid_pos[0]
    
    # Case 3: already correct
    return current_pos

token_pos_copy["pos"] = token_pos_copy.apply(fix_pos, axis=1)
```

```python
def cleanPOSDF(df):
    df = df[df['token_lower'] != '-'] 
    # remove rows with hyphen token
    df = df.replace(r"^\s*$", np.nan, regex=True).dropna(subset=["token_lower"])
    # remove rows with space or nan
    return df

def getPOS(df, rows=35):
    cleanPOSDF(df)
    res_df = pd.DataFrame()
    res_df['noun'] = df[df['pos'] == 'NOUN']['token_lower'].value_counts()[:rows].index.to_list()
    res_df['verb'] = df[df['pos'] == 'VERB']['token_lower'].value_counts()[:rows].index.to_list()
    res_df['pron'] = df[df['pos'] == 'PRON']['token_lower'].value_counts()[:rows].index.to_list()
    res_df['adj'] = df[df['pos'] == 'ADJ']['token_lower'].value_counts()[:rows].index.to_list()
    res_df['adv'] = df[df['pos'] == 'ADV']['token_lower'].value_counts()[:rows].index.to_list()
    res_df['propn'] = df[df['pos'] == 'PROPN']['token_lower'].value_counts()[:rows].index.to_list()

    return res_df
```

```python
def clean_text(text: str) -> str:
    """Lowercase, remove punctuation, lemmatize, remove stopwords."""
    text = text.lower()
    text = re.sub(r'[^\w\s-]', ' ', text) # remove punctuation excluding hyphen
    text = re.sub(r'\n', ' ', text) # replace newline wit space
    text = re.sub(r' +', ' ', text) # remove extraspaces
    text = text.strip() # remove leading and trailing whitespace
    
    return text
```

```python
joined_tokens['processed'] = joined_tokens['token'].apply(clean_text)
```
